## Q2: Building Markov Transition Matrix

You have learned the key differences between a simple dinucleotide frequency model and Markov models.

**Task:**  
Given an input DNA sequence, build a **Markov transition matrix** that represents the transition probabilities between nucleotides (A, C, G, T).

**Expected Output:**  
A 4x4 matrix showing transition probabilities from one nucleotide to another (e.g., from A to C, G, T, etc.).

### Markov Transition Matrix Construction

This code builds a **Markov transition matrix** from a given DNA sequence. It performs the following steps:

1. **Initialization:**
   - Defines the four nucleotide states: `A`, `C`, `G`, and `T`.
   - Initializes a nested dictionary to store transition counts between nucleotides.

2. **Counting Transitions:**
   - Iterates through the sequence and counts how often each nucleotide is followed by another.
   - Ignores any invalid characters if present.

3. **Calculating Probabilities:**
   - For each nucleotide, the transition counts are normalized (i.e., divided by the total number of transitions from that nucleotide) to convert them into probabilities.

4. **Output:**
   - The final Markov transition matrix is printed using `pandas.DataFrame`, with:
     - **Rows** representing the current nucleotide,
     - **Columns** representing the next nucleotide.


```python
import pandas as pd

dna_sequence = "AGTTCAGTTATTAATCTGGCTTATATACTGACTACTATATCTCTTATATATACT"

nucleotides = ['A', 'C', 'G', 'T']

# Initialize a nested dictionary to store transition counts between nucleotides
transition_counts = {base: {next_base: 0 for next_base in nucleotides} for base in nucleotides}

for idx in range(len(dna_sequence) - 1):
    curr_base = dna_sequence[idx]
    next_base = dna_sequence[idx + 1]

    if curr_base in nucleotides and next_base in nucleotides:
        transition_counts[curr_base][next_base] += 1
    else:
        print(f"Skipped invalid base transition: '{curr_base}' → '{next_base}'")

# Initialize a dictionary to store probabilities for the Markov transition matrix
markov_probabilities = {from_base: {} for from_base in nucleotides}

# Convert counts to probabilities (row-wise normalization)
for from_base in nucleotides:
    total_transitions = sum(transition_counts[from_base].values())

    for to_base in nucleotides:
        if total_transitions > 0:
            markov_probabilities[from_base][to_base] = transition_counts[from_base][to_base] / total_transitions
        else:
            markov_probabilities[from_base][to_base] = 0.0

# Create a DataFrame for better visualization
transition_df = pd.DataFrame(markov_probabilities)

# Display the resulting transition matrix
print("Markov Transition Matrix:\n(rows: current base, columns: next base)")
print(transition_df)
```
