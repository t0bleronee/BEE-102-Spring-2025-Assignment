## Q3: Multi-line sequence fasta to single-line fasta file



File Handling:
- Opens the input FASTA file (multiline_input.fasta) for reading and creates an output file (singleline_output.fasta) for writing the processed content.

Reading and Writing Logic:

- Iterates through each line in the input file.

- If the line starts with '>', it is treated as a header line:

- If it’s not the first header, a newline is added before writing the new header.

- The header is written as-is, followed by a newline.

- If the line is part of the sequence, it is stripped of newline characters and appended directly (i.e., concatenated) to ensure the full sequence appears on one line.

End of File:
- Adds a final newline to ensure the file ends cleanly.

Output:
- The resulting file singleline_output.fasta will have each header on one line and its corresponding sequence on the next line, uninterrupted by line breaks.

```python
input_fasta = "multiline_input.fasta"
output_fasta = "singleline_output.fasta"

with open(input_fasta, 'r') as infile, open(output_fasta, 'w') as outfile:

    # Flag to track if we’ve already seen the first header
    first_header_processed = False

    # Iterate through each line in the input FASTA file
    for line in infile:
        clean_line = line.strip('\n')

        if clean_line.startswith('>'):

            if first_header_processed:
                outfile.write('\n')

            # Write the header line to the output file
            outfile.write(clean_line + '\n')

            # Mark that we've now processed at least one header
            first_header_processed = True
        else:
            outfile.write(clean_line.strip())

    # Ensure the file ends with a newline character
    outfile.write('\n')

print(f"The converted output is saved as '{output_fasta}'.")
```
